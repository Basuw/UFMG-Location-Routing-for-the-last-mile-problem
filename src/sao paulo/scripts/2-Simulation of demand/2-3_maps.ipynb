```python
import numpy as np
import pandas as pd
import folium
from branca.colormap import linear
```

```python
df = pd.read_csv('../../results/utils/df_clients_grids.csv')[['Order Postal Code', 'Grid', 'maxLat', 'minLat', 'maxLong', 'minLong']]
df1 = df.groupby(by=['Grid', 'maxLat', 'minLat', 'maxLong', 'minLong'], group_keys=False).count().reset_index()
```

```python
# Create map centered on initial location
map_ = folium.Map(location=[-23.802923, -46.728143], zoom_start=12)

# Normalize the 'quantity' column between 0 and 1
min_quantidade = df1['Order Postal Code'].min()
max_quantidade = df1['Order Postal Code'].max()

# Define color scale
color_scale = linear.YlOrRd_09.scale(min_quantidade,max_quantidade)

# Iterate over each row of the dataframe
for index, row in df1.iterrows():
    # Extract info from each row
    cell = row['Grid']
    max_lat = row['maxLat']
    min_lat = row['minLat']
    max_long = row['maxLong']
    min_long = row['minLong']
    quantidade_normalizada = row['Order Postal Code']
    
    # Compute color based on quantity
    cor = color_scale(quantidade_normalizada)
    
    # Draw rectangle on map for the grid cell with computed color
    folium.Rectangle(
        bounds=[(min_lat, min_long), (max_lat, max_long)],
        color='gray',
        weight=1,
        fill=True,
        fill_color=cor,
        fill_opacity = 0.8, 
        tooltip = row['Order Postal Code']
    ).add_to(map_)

color_scale.caption = 'Clients per cluster'
color_scale.add_to(map_)

sw = [df1['minLat'].min(), df1['minLong'].min()]
ne = [df1['maxLat'].max(), df1['maxLong'].max()]
bounds = [sw, ne]
map_.fit_bounds(bounds=bounds)

# Display map
folium.TileLayer('cartodbpositron').add_to(map_)
map_.save("../../results/maps/demand_per_square.html")

map_
```
